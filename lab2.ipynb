```python
import pandas as pd
import numpy as np
```

```python
# Таблица 1: Продажи
sales = pd.DataFrame({
'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone','Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
'Price': [50000, 30000, 20000, 50000, np.nan,
50000, 20000, 30000, 50000, np.nan],
'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb','Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']})
# Таблица 2: Информация о товарах
products = pd.DataFrame({
'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})
print("Таблица продаж:")
print(sales)
print("\nТаблица товаров:")
print(products)
```

**ЗАДАНИЕ 1.Найти среднее значение Quantity для каждого продукта**

```python
print(sales.groupby('Product')['Quantity'].mean())
```

**ЗАДАНИЕ 2.  Найти продукт с максимальной общей выручкой (сумма Quantity * Price )**

```python
sales['Revenue']= sales['Price']*sales['Quantity']
print(sales.groupby('Product')['Revenue'].sum().idxmax())
```

**ЗАДАНИЕ 3. Заполнить пропуски в Region методом forward fill ( .fillna(method='ffill') )**

```python
sales['Region'] = sales['Region'].ffill()
print(sales['Region'])
```

**ЗАДАНИЕ 4. Отсортировать данные по OrderID в обратном порядке**

```python
print(sales.sort_values('OrderID',ascending = False))
```

**ЗАДАНИЕ 5. Выполнить outer join таблиц sales и products**

```python
print(pd.merge(sales, products, on='Product', how='outer'))
```
